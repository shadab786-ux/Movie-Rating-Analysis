## Genre Distribution:

```python
genre_counts = df['Genre'].value_counts()

plt.figure(figsize=(10,6))

genre_counts.plot(kind='bar')

plt.title("Movie Distribution by Genre")
plt.xlabel("Genre")
plt.ylabel("Number of Movies")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

###### The chart shows the distribution of movies across different genres. The most common genre has the highest number of movies and in this chart action genre has the highest number of movies

## Genre Distribution with Pie Chart:

```python
genre_counts = df['Genre'].value_counts()

plt.figure(figsize=(8,8))

plt.pie(
    genre_counts.values,
    labels=genre_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Genre Distribution of Movies")

plt.show()
```

## Average Rating by Genre:

```python
avg_rating_genre = df.groupby('Genre')['Rating'].mean().sort_values(ascending=False)

plt.figure(figsize=(10,6))

avg_rating_genre.plot(kind='bar')

plt.title("Average Rating by Genre")
plt.xlabel("Genre")
plt.ylabel("Average Rating")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

## Movies Released by Year:

```python
movies_by_year = df.groupby('Release_Year')['Title'].count()

plt.figure(figsize=(12,6))

movies_by_year.plot(kind='line', marker='o')

plt.title("Number of Movies Released by Year")
plt.xlabel("Year")
plt.ylabel("Number of Movies")

plt.grid(True)

plt.tight_layout()
plt.show()
```

## Most Popular Year:

```python
popular_year = df.groupby('Release_Year')['Votes'].mean().sort_values(
    ascending=False
)

plt.figure(figsize=(12,6))

popular_year.head(10).plot(kind='bar',color="skyblue")

plt.title("Top 10 Most Popular Years")
plt.xlabel(" Releas Year")
plt.ylabel("Average Popularity")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
most_popular_year = popular_year.idxmax()

print("Most Popular Year:", most_popular_year)
```

## Rating Distribution:

```python
plt.figure(figsize=(10,6))

plt.hist(
    df['Rating'],
    bins=10,
    edgecolor='black'
)

plt.title("Movie Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Number of Movies")

plt.tight_layout()
plt.show()
```

## Rating Trend Over Years:

```python
year_rating = df.groupby('Release_Year')['Rating'].mean()

plt.figure(figsize=(12,6))

year_rating.plot(kind='line', marker='o')

plt.title("Average Movie Rating Over Years")
plt.xlabel("Year")
plt.ylabel("Average Rating")

plt.grid(True)

plt.tight_layout()
plt.show()
```

## Top 10 Highest Rated Movies:

```python
top_movies = df.sort_values(
    by='Rating',
    ascending=False
).head(10)

plt.figure(figsize=(10,6))

plt.barh(
    top_movies['Title'],
    top_movies['Rating']
)

plt.title("Top 10 Highest Rated Movies")
plt.xlabel("Rating")
plt.ylabel("Movie")

plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```

## Rating vs Popularity

```python
plt.figure(figsize=(10,6))

plt.scatter(
    df['Rating'],
    df['Votes'],
    alpha=0.6
)

plt.title("Rating vs Popularity")
plt.xlabel("Rating")
plt.ylabel("Popularity")

plt.tight_layout()
plt.show()
```

## Correlation:

```python
correlation = df.select_dtypes(include=np.number).corr()

print(correlation)
```

### Heatmap:

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")

plt.tight_layout()
plt.show()
```
